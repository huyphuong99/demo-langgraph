```python
# How It Works

# LangGraph operates based on a modular architecture that separates the concerns of graph creation, manipulation, and visualization. 
# This design ensures flexibility and scalability, making it suitable for a wide range of use cases.

# Core Components:
# 1. **GraphBuilder**: Responsible for creating and initializing graphs.
# 2. **GraphManipulator**: Provides methods to modify and analyze graphs.
# 3. **GraphVisualizer**: Handles the rendering and visualization of graphs.

# Workflow:
# 1. **Initialize**: Use the GraphBuilder to create a new graph.
# 2. **Manipulate**: Modify the graph using GraphManipulator methods (e.g., add nodes, edges).
# 3. **Visualize**: Render the graph using GraphVisualizer for insights.

# Example Code:

# Importing LangGraph library
from langgraph.graph import StateGraph
from langgraph import GraphManipulator, GraphVisualizer

# Step 1: Initialize the graph
graph = StateGraph().create_graph(directed=True)  # Creating a directed graph

# Step 2: Manipulate the graph
manipulator = GraphManipulator(graph)
manipulator.add_node("A")  # Adding a node
manipulator.add_node("B")  # Adding another node
manipulator.add_edge("A", "B")  # Adding a directed edge from A to B

# Step 3: Visualize the graph
visualizer = GraphVisualizer(graph)
visualizer.render()  # Render the graph for visualization
```

```python
from typing import TypedDict, Annotated, List
from langgraph.graph import StateGraph, END
from langchain_openai import ChatOpenAI

# Định nghĩa cấu trúc trạng thái
class MathProblemState(TypedDict):
    problem: str
    steps: List[str]
    solution: str
    complete: bool

# Khởi tạo mô hình ngôn ngữ
llm = ChatOpenAI(model="gpt-4")

# Các nút trong đồ thị
def analyze_problem(state: MathProblemState) -> MathProblemState:
    """Phân tích bài toán và lập kế hoạch giải quyết"""
    response = llm.invoke(f"Hãy phân tích bài toán sau và đề xuất cách tiếp cận: {state['problem']}")
    new_steps = state["steps"] + [f"Phân tích: {response.content}"]
    return {"steps": new_steps}

def solve_step_by_step(state: MathProblemState) -> MathProblemState:
    """Giải bài toán từng bước một"""
    response = llm.invoke(f"""
    Bài toán: {state['problem']}
    Phân tích hiện tại: {state['steps'][-1]}
    Hãy giải từng bước một cách rõ ràng.
    """)
    new_steps = state["steps"] + [f"Giải pháp từng bước: {response.content}"]
    return {"steps": new_steps}

def verify_solution(state: MathProblemState) -> MathProblemState:
    """Kiểm tra lại giải pháp"""
    response = llm.invoke(f"""
    Bài toán: {state['problem']}
    Giải pháp: {state['steps'][-1]}
    Hãy kiểm tra lại giải pháp và đưa ra kết quả cuối cùng.
    """)
    return {
        "solution": response.content,
        "complete": True
    }

def should_continue(state: MathProblemState) -> str:
    """Kiểm tra xem đã hoàn thành chưa"""
    return "end" if state["complete"] else "continue"

# Xây dựng đồ thị
builder = StateGraph(MathProblemState)
builder.add_node("analyze", analyze_problem)
builder.add_node("solve", solve_step_by_step)
builder.add_node("verify", verify_solution)

# Định nghĩa luồng đi
builder.add_edge("analyze", "solve")
builder.add_edge("solve", "verify")
builder.add_conditional_edges(
    "verify",
    should_continue,
    {
        "continue": "analyze",  # Quay lại phân tích nếu chưa hoàn thành
        "end": END
    }
)

# Biên dịch đồ thị
math_solver = builder.compile()

# Sử dụng
result = math_solver.invoke({
    "problem": "Tìm các giá trị của x thỏa mãn phương trình x^2 - 5x + 6 = 0",
    "steps": [],
    "solution": "",
    "complete": False
})

print(result["solution"])
```

```python
from typing import TypedDict, List, Dict
from langgraph.graph import StateGraph, END
from langchain_openai import ChatOpenAI

# Định nghĩa cấu trúc trạng thái
class CustomerSupportState(TypedDict):
    chat_history: List[Dict[str, str]]
    user_query: str
    context: Dict[str, any]
    response: str
    needs_escalation: bool

# Các nút trong đồ thị
def analyze_query(state: CustomerSupportState) -> CustomerSupportState:
    """Phân tích câu hỏi của người dùng"""
    llm = ChatOpenAI(model="gpt-3.5-turbo")
    response = llm.invoke(f"""
    Dựa vào lịch sử trò chuyện và câu hỏi hiện tại, xác định xem người dùng đang hỏi về vấn đề gì?
    Lịch sử: {state['chat_history']}
    Câu hỏi hiện tại: {state['user_query']}
    """)
    
    return {"context": {"topic": response.content}}

def retrieve_info(state: CustomerSupportState) -> CustomerSupportState:
    """Truy xuất thông tin liên quan"""
    topic = state["context"].get("topic", "")
    # Giả lập truy xuất thông tin từ cơ sở dữ liệu
    if "đơn hàng" in topic.lower():
        info = "Thông tin về đơn hàng: Bạn có thể kiểm tra trạng thái đơn hàng tại trang cá nhân."
    elif "hoàn tiền" in topic.lower():
        info = "Chính sách hoàn tiền: Hoàn tiền trong vòng 7 ngày sau khi nhận hàng."
    else:
        info = "Thông tin chung về sản phẩm và dịch vụ của chúng tôi."
    
    context = state["context"]
    context["relevant_info"] = info
    return {"context": context}

def generate_response(state: CustomerSupportState) -> CustomerSupportState:
    """Tạo câu trả lời cho người dùng"""
    llm = ChatOpenAI(model="gpt-3.5-turbo")
    chat_history = state["chat_history"]
    relevant_info = state["context"].get("relevant_info", "")
    
    response = llm.invoke(f"""
    Dựa vào lịch sử trò chuyện, thông tin liên quan, và câu hỏi hiện tại, hãy tạo câu trả lời hữu ích.
    Lịch sử: {chat_history}
    Thông tin liên quan: {relevant_info}
    Câu hỏi hiện tại: {state['user_query']}
    """)
    
    return {"response": response.content}

def check_escalation(state: CustomerSupportState) -> CustomerSupportState:
    """Kiểm tra xem có cần chuyển tiếp đến nhân viên không"""
    llm = ChatOpenAI(model="gpt-3.5-turbo")
    response = llm.invoke(f"""
    Dựa vào câu hỏi và câu trả lời, xác định xem vấn đề này có cần chuyển tiếp cho nhân viên hỗ trợ không? 
    Trả lời "Có" hoặc "Không".
    Câu hỏi: {state['user_query']}
    Câu trả lời: {state['response']}
    """)
    
    needs_escalation = "có" in response.content.lower()
    return {"needs_escalation": needs_escalation}

def should_escalate(state: CustomerSupportState) -> str:
    """Quyết định luồng đi tiếp theo"""
    return "escalate" if state["needs_escalation"] else "end"

# Xây dựng đồ thị
builder = StateGraph(CustomerSupportState)
builder.add_node("analyze", analyze_query)
builder.add_node("retrieve", retrieve_info)
builder.add_node("generate", generate_response)
builder.add_node("check_escalation", check_escalation)

# Định nghĩa luồng đi
builder.add_edge("analyze", "retrieve")
builder.add_edge("retrieve", "generate")
builder.add_edge("generate", "check_escalation")
builder.add_conditional_edges(
    "check_escalation",
    should_escalate,
    {
        "escalate": "escalate_to_human",  # Nút giả định
        "end": END
    }
)

# Biên dịch đồ thị
customer_support = builder.compile()
```

```python
from typing import TypedDict, List
from langgraph.graph import StateGraph, END
from langchain_openai import ChatOpenAI

# Định nghĩa cấu trúc trạng thái
class ResearchState(TypedDict):
    query: str
    sub_questions: List[str]
    findings: List[dict]
    summary: str
    complete: bool

# Các nút trong đồ thị
def break_down_query(state: ResearchState) -> ResearchState:
    """Chia câu hỏi lớn thành các câu hỏi nhỏ hơn"""
    llm = ChatOpenAI(model="gpt-4")
    response = llm.invoke(f"""
    Chia câu hỏi lớn này: "{state['query']}" 
    thành 3-5 câu hỏi nhỏ hơn để nghiên cứu. Liệt kê từng câu hỏi trên một dòng.
    """)
    
    sub_questions = [q.strip() for q in response.content.split("\n") if q.strip()]
    return {"sub_questions": sub_questions}

def research_question(state: ResearchState) -> ResearchState:
    """Nghiên cứu từng câu hỏi nhỏ"""
    llm = ChatOpenAI(model="gpt-4")
    findings = state.get("findings", [])
    
    # Nếu đã nghiên cứu tất cả các câu hỏi
    if len(findings) >= len(state["sub_questions"]):
        return {"complete": True}
    
    # Lấy câu hỏi tiếp theo cần nghiên cứu
    question_index = len(findings)
    current_question = state["sub_questions"][question_index]
    
    # Giả lập việc tìm kiếm thông tin
    response = llm.invoke(f"""
    Hãy nghiên cứu và trả lời câu hỏi sau: "{current_question}"
    Đưa ra câu trả lời dựa trên kiến thức của bạn, càng chi tiết càng tốt.
    """)
    
    new_finding = {
        "question": current_question,
        "answer": response.content
    }
    
    updated_findings = findings + [new_finding]
    return {"findings": updated_findings}

def synthesize_results(state: ResearchState) -> ResearchState:
    """Tổng hợp kết quả từ tất cả các câu hỏi"""
    llm = ChatOpenAI(model="gpt-4")
    findings_text = "\n\n".join([
        f"Câu hỏi: {f['question']}\nTrả lời: {f['answer']}" 
        for f in state["findings"]
    ])
    
    response = llm.invoke(f"""
    Dựa trên các phát hiện sau đây từ nghiên cứu về "{state['query']}", 
    hãy viết một bản tổng hợp toàn diện, có cấu trúc rõ ràng:
    
    {findings_text}
    """)
    
    return {"summary": response.content}

def check_complete(state: ResearchState) -> str:
    """Kiểm tra xem có cần tiếp tục nghiên cứu hay không"""
    return "synthesize" if state.get("complete", False) else "continue_research"

# Xây dựng đồ thị
builder = StateGraph(ResearchState)
builder.add_node("break_down", break_down_query)
builder.add_node("research", research_question)
builder.add_node("synthesize", synthesize_results)

# Định nghĩa luồng đi
builder.add_edge("break_down", "research")
builder.add_conditional_edges(
    "research",
    check_complete,
    {
        "continue_research": "research",  # Tiếp tục nghiên cứu câu hỏi tiếp theo
        "synthesize": "synthesize"  # Chuyển sang tổng hợp khi đã nghiên cứu xong
    }
)
builder.add_edge("synthesize", END)

# Biên dịch đồ thị
researcher = builder.compile()

# Sử dụng
result = researcher.invoke({
    "query": "Tác động của AI đối với thị trường lao động trong thập kỷ tới",
    "sub_questions": [],
    "findings": [],
    "summary": "",
    "complete": False
})

print(result["summary"])
```
